```python
import geopandas as gpd
import pandas as pd
from shapely import Polygon, Point
from shapely import plotting
```

```python
pg = Polygon([(0,0),(4,0),(2,3.5),(0,0)])
sq = Polygon([(2,2),(3,2),(3,4),(2,4),(2,2)])
```

```python
pg
```

```python
sq
```

```python
pg.intersects(sq)
```

```python
pg.intersection(sq)
```

```python
pg.union(sq)
```

```python
plotting.plot_polygon(pg)
plotting.plot_polygon(sq,color='r')
```
